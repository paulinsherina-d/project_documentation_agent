```python
from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet

pdf_path = "AI_Project_Documentation.pdf"

doc = SimpleDocTemplate(pdf_path, pagesize=A4)
styles = getSampleStyleSheet()

content = []

content.append(Paragraph("AI Project Documentation Agent", styles["Title"]))
content.append(Spacer(1, 20))

sections = [
    ("1. Project Objective",
     "The AI Project Documentation Agent is an intelligent assistant that helps users understand and retrieve information from AI project documentation. It uses Retrieval Augmented Generation (RAG) to find relevant information from project documents and generate useful answers."),

    ("2. Problem Statement",
     "AI project documents can contain large amounts of technical information. Finding specific information manually can take time. The proposed agent provides a conversational interface where users can ask questions about the project documentation."),

    ("3. Technologies Used",
     "The project uses Python, Google Colab, LangChain, Groq, HuggingFace Embeddings, FAISS, PyPDF, and Gradio."),

    ("4. System Architecture",
     "The system loads a project PDF, extracts its text, divides the text into smaller chunks, converts the chunks into embeddings, stores them in a FAISS vector database, retrieves relevant information, and sends the retrieved context to a Groq language model."),

    ("5. Main Modules",
     "The main modules are PDF loading, text splitting, embeddings generation, FAISS vector database, document retriever, RAG tool, language model, agent, memory, and Gradio user interface."),

    ("6. Working Methodology",
     "First, the project documentation is loaded from a PDF. The document is divided into smaller chunks using a recursive character text splitter. HuggingFace embeddings convert the text chunks into numerical vectors. FAISS stores these vectors and allows relevant information to be retrieved. The retrieved information is provided to the Groq language model to generate an answer."),

    ("7. RAG Process",
     "Retrieval Augmented Generation combines document retrieval with a language model. When a user asks a question, the retriever searches the vector database for relevant document chunks. These chunks are then used as context for generating the answer."),

    ("8. Advantages",
     "The agent provides quick access to project information, reduces manual document searching, supports natural language questions, and can be extended to support multiple project documents."),

    ("9. Applications",
     "The AI Project Documentation Agent can be used by students, developers, project teams, researchers, and organizations to understand technical project documentation."),

    ("10. Future Enhancements",
     "Future versions can support multiple PDF documents, better conversation memory, project cost estimation, document summarization, multilingual questions, authentication, and a web-based deployment."),

    ("11. Conclusion",
     "The AI Project Documentation Agent demonstrates how RAG, vector databases, embeddings, and large language models can be combined to create a useful document-question-answering system.")
]

for title, text in sections:
    content.append(Paragraph(title, styles["Heading2"]))
    content.append(Paragraph(text, styles["BodyText"]))
    content.append(Spacer(1, 12))

doc.build(content)

print("PDF created successfully!")
print("File:", pdf_path)
```

```python
from langchain_community.document_loaders import PyPDFLoader

pdf_path = "/content/AI_Project_Documentation.pdf"

loader = PyPDFLoader(pdf_path)
documents = loader.load()

print("PDF loaded successfully!")
print("Number of pages:", len(documents))
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = text_splitter.split_documents(documents)

print("PDF divided into chunks!")
print("Number of chunks:", len(chunks))
$0
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embeddings model loaded successfully!")
```

```python
from langchain_community.vectorstores import FAISS

vectorstore = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)

print("FAISS vector database created successfully!")
```

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

print("Retriever created successfully!")
```

```python
question = "What is the objective of the AI Project Documentation Agent?"

results = retriever.invoke(question)

print("Retrieved information:\n")

for i, doc in enumerate(results, 1):
    print(f"--- Result {i} ---")
    print(doc.page_content)
    print()
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are an AI Project Documentation Agent.

Answer the user's question using ONLY the information provided
in the project documentation below.

If the answer is not present in the documentation, say:
"I could not find that information in the project documentation."

Project Documentation:
{context}

User Question:
{question}
""")

print("LangChain prompt created successfully!")
```

```python
from langchain_core.tools import tool

@tool
def documentation_search(question: str) -> str:
    """Search the AI Project Documentation and return relevant information."""

    results = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in results
    )

    return context

print("RAG tool created successfully!")
```

```python
from langchain_core.tools import tool

@tool
def budget_calculator(items: str) -> str:
    """
    Calculate the total project budget.
    Enter items as: item=price, item=price
    Example: laptop=50000, hosting=5000
    """
    try:
        total = 0

        for item in items.split(","):
            name, price = item.split("=")
            total += float(price.strip())

        return f"Estimated total project budget: ₹{total:,.2f}"

    except Exception:
        return "Please enter items like: laptop=50000, hosting=5000"

print("Budget tool created successfully!")
```

```python
print(budget_calculator.invoke("laptop=50000, hosting=5000, software=3000"))
```

```python
def ask_agent(question):
    results = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in results
    )

    final_prompt = prompt.invoke({
        "context": context,
        "question": question
    })

    response = llm.invoke(final_prompt)

    return response.content


answer = ask_agent(
    "What are the technologies used in this project?"
)

print("Agent Answer:")
print(answer)
```

```python
from langchain_core.messages import HumanMessage, AIMessage

chat_history = []

print("Memory created successfully!")
```

```python
def agent_with_memory(question):
    global chat_history

    # Search the project documentation
    results = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in results
    )

    # Include previous conversation
    history = "\n".join(
        f"User: {h['user']}\nAgent: {h['agent']}"
        for h in chat_history
    )

    final_prompt = prompt.invoke({
        "context": context,
        "question": f"""
Previous conversation:
{history}

Current question:
{question}
"""
    })

    response = llm.invoke(final_prompt)
    answer = response.content

    # Save conversation
    chat_history.append({
        "user": question,
        "agent": answer
    })

    return answer


print("Agent with memory created successfully!")
```

```python
print(agent_with_memory("What is the objective of this project?"))
```

```python
print(agent_with_memory("What technologies are used in it?"))
```

```python
import gradio as gr

def chat_with_agent(message, history):
    return agent_with_memory(message)

demo = gr.ChatInterface(
    fn=chat_with_agent,
    title="🤖 AI Project Documentation Agent",
    description="Ask questions about the AI Project Documentation.",
    textbox=gr.Textbox(
        placeholder="Ask something about the project...",
        container=True
    )
)

demo.launch(share=True)
```
